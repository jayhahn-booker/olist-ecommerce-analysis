## Question 9 — Delivery Time Analysis
Calculating delivery duration distribution, regional performance,
and trends over time to identify operational inefficiencies.

```python
# Calculate delivery time in days
# Convert relevant columns to datetime if they are not already
df_orders['order_delivered_customer_date'] = pd.to_datetime(df_orders['order_delivered_customer_date'])
df_orders['order_purchase_timestamp'] = pd.to_datetime(df_orders['order_purchase_timestamp'])
df_orders['delivery_days'] = (df_orders['order_delivered_customer_date'] - df_orders['order_purchase_timestamp']).dt.days
print(df_orders['delivery_days'].describe())
```

```python
#Remove outliers for cleaner analysis
#Filter to reasonable delivery window of under 60 days
delivery_clean = df_orders[(df_orders['delivery_days'] >= 1) & (df_orders['delivery_days'] <= 60)]['delivery_days']

print(f"Orders within 60 day window: {len(delivery_clean)}")
print(f"Orders excluded as outliers: {len(df_orders.dropna(subset={'delivery_days'})) - len(delivery_clean)}")
```

```python
#Merge delivery data with customer state information
delivery_by_state =df_orders.merge(df_customers[['customer_id', 'customer_state']], on= 'customer_id', how= 'left')

#Filter clean delivery window
delivery_by_state = delivery_by_state[(delivery_by_state['delivery_days'] >= 1) & (delivery_by_state['delivery_days']<= 60)]

#Calculate average delivery time by state
state_delivery = delivery_by_state.groupby('customer_state')['delivery_days'].agg([ 'mean', 'median', 'count']).round(2).sort_values('mean', ascending=False)
print(state_delivery)
```

```python
#Calculate best and worst performing states summary
print("Top Fastest Delivery States")
print("=" * 45)
print(state_delivery.tail(5)[['mean', 'median', 'count']])

print("\nTop 5 Slowest Delivery States")
print("=" * 45)
print(state_delivery.head(5)[['mean', 'median', 'count']])
```

```python
#Identify late deliveries
# Ensure both columns are datetime objects
df_orders['order_delivered_customer_date'] = pd.to_datetime(df_orders['order_delivered_customer_date'])
df_orders['order_estimated_delivery_date'] = pd.to_datetime(df_orders['order_estimated_delivery_date'])
df_orders['is_late']= ( df_orders['order_delivered_customer_date'] > df_orders['order_estimated_delivery_date'])

#Calculate days late (negative means early)
df_orders['days_late'] = (df_orders['order_delivered_customer_date'] - df_orders['order_estimated_delivery_date']).dt.days

# Late delivery summary
total_delivered = df_orders['is_late'].notna().sum()
late_orders = df_orders['is_late'].sum()
late_percentage = (late_orders / total_delivered) * 100

print(f"Total delivered orders: {total_delivered}")
print(f"Late orders: {late_orders}")
print(f"Late delivery percentage: {late_percentage:.2F}%")
```

```python
# Analyzenhow late the late orders actually were
late_orders_df = df_orders[df_orders['is_late'] == True]

print("Late Delivery Magnitude Analysis")
print("=" * 45)
print(late_orders_df['days_late'].describe().round(2))
```

```python
# Merge orders with reviews
orders_reviews = df_orders.merge( df_reviews[['order_id', 'review_score']], on='order_id', how='left')

# Compare average review scores for late vs on time deliveries
delivery_review = orders_reviews.groupby('is_late')['review_score'].agg(['mean', 'median', 'count']).round(2)
delivery_review.index =['On Time', 'Late']

print("Average Review Score By Delivery Performance")
print("=" * 45)
print(delivery_review)
```

```python
# Calculate revenue at risk from late deliveries
late_with_payments = orders_reviews.merge( df_payments.groupby('order_id')['payment_value'].sum().reset_index(), on='order_id', how='left')
late_revenue = late_with_payments[late_with_payments['is_late'] == True]['payment_value'].sum()
total_revenue = late_with_payments['payment_value'].sum()
late_revenue_percentage = (late_revenue / total_revenue) * 100

print(f"Total Revenue: R${total_revenue:,.2f}")
print(f"Revenue from Late Orders: R${late_revenue:,.2f}")
print(f"Percentage of Revenue at Risk: {late_revenue_percentage:.2f}")
```

## Question 13 — Top Seller Performance Analysis
Building complete seller performance profiles combining revenue,
customer satisfaction, and product category data to identify
revenue assets versus customer satisfaction liabilities.

```python
import pandas as pd
from google.colab import drive
drive.mount('/content/drive')

# Re-loading necessary DataFrames to ensure they are defined in case of kernel reset or out-of-order execution.
# These are typically loaded in an earlier cell (XGyKsWEbelnq).
df_order_items = pd.read_csv('/content/drive/MyDrive/SQL Project (Brazil Ecommerce) Folder/Raw Tables/olist_order_items_dataset.csv')
df_orders = pd.read_csv('/content/drive/MyDrive/SQL Project (Brazil Ecommerce) Folder/Raw Tables/olist_orders_dataset.csv')
df_reviews = pd.read_csv('/content/drive/MyDrive/SQL Project (Brazil Ecommerce) Folder/Raw Tables/olist_order_reviews_dataset.csv')
df_products = pd.read_csv('/content/drive/MyDrive/SQL Project (Brazil Ecommerce) Folder/Raw Tables/olist_products_dataset.csv')
df_category = pd.read_csv('/content/drive/MyDrive/SQL Project (Brazil Ecommerce) Folder/Raw Tables/product_category_name_translation.csv')

#Step 1 - Merge order items with orders
seller_data = df_order_items.merge(df_orders[['order_id']], on='order_id', how='left')

#Step 2 - Merge with reviews
seller_data = seller_data.merge(df_reviews[['order_id','review_score']], on='order_id', how='left')

#Step 3 - Merge with products
seller_data = seller_data.merge(df_products[['product_id', 'product_category_name']], on='product_id', how='left')

#Step 4 - Merge with category translation
seller_data = seller_data.merge(df_category[['product_category_name', 'product_category_name_english']], on='product_category_name', how='left')

print(f"Merged dataset shape: {seller_data.shape}")
print(seller_data.head())
```

```python
#Calculate top 10 sellers by revenue
top_10_sellers = seller_data.groupby('seller_id').agg(total_revenue=('price', 'sum'),
    avg_review_score=('review_score', 'mean'), total_orders=('order_id', 'count')
    ).round(2).sort_values('total_revenue', ascending=False).head(10).reset_index()

print("Top 10 Sellers by Revenue")
print("=" * 65)
print(top_10_sellers)
```

```python
# Detailed view of revenue vs satisfaction relationship
print("Top 10 Sellers - Revenue vs Satisfaction Analysis")
print("=" * 65)

for idx, row in top_10_sellers.iterrows():
  if row['avg_review_score']>= 4.0:
    status = 'Asset'
  elif row['avg_review_score']>= 3.5:
    status = 'Monitor'
  else:
    status = 'Liability'
  print(f"Rank {idx+1} | Revenue: R${row['total_revenue']:,.2f} |"
        f"Score: {row['avg_review_score']} | {status}")
```

```python
# Deep dive on liability seller
liability_seller =top_10_sellers.iloc[3]['seller_id']

# Get their product category breakdown
seller_categories = seller_data[seller_data['seller_id'] == liability_seller].groupby(
    'product_category_name_english').agg(total_revenue =('price','sum'),
    avg_review_score=('review_score', 'mean'), order_count=('order_id', 'count')
    ).round(2).sort_values('total_revenue', ascending=False)

print(f"Product Category Breakdown for Liability Seller")
print("=" * 45)
print (seller_categories)
```

```python
# Deep dive on monitor seller
monitor_seller = top_10_sellers.iloc[1]['seller_id']

monitor_categories = seller_data[seller_data['seller_id'] == monitor_seller].groupby(
    'product_category_name_english').agg(
    total_revenue=('price', 'sum'),
    avg_review_score=('review_score', 'mean'),
    order_count=('order_id', 'count')
).round(2).sort_values('total_revenue', ascending=False)

print(f"Product Category Breakdown for Monitor Seller")
print("=" * 65)
print(monitor_categories)
```

```python
# Check for uncategorized orders from monitor seller
monitor_full = seller_data[seller_data['seller_id'] == monitor_seller]

print(f"Total orders: {len(monitor_full)}")
print(f"Orders with null category: {monitor_full['product_category_name_english'].isna().sum()}")
print(f"Orders with null review: {monitor_full['review_score'].isna().sum()}")
print(f"\nOverall average review: {monitor_full['review_score'].mean():.2f}")
print(f"Review score distribution:")
print(monitor_full['review_score'].value_counts().sort_index())
```

## Question 15 — Revenue Concentration Analysis
Analyzing revenue distribution across the seller base to identify
concentration risk and test the Pareto Principle in this dataset.

```python
# Calculate total revenue per seller
seller_revenue = df_order_items.groupby('seller_id')['price'].sum().reset_index()
seller_revenue.columns = ['seller_id','total_revenue']
seller_revenue = seller_revenue.sort_values('total_revenue', ascending=False).reset_index(drop=True)

print(f"Total sellers: {len(seller_revenue)}")
print(f"Total revenue: R${seller_revenue['total_revenue'].sum():,.2f}")
print(seller_revenue.head(10))
```

```python
#Calculate Top 10% threshold
total_sellers = len(seller_revenue)
top_10_percent_threshold = int(total_sellers *0.10)
print(f"Total sellers: {total_sellers}")
print(f"Top 10% threshold: {top_10_percent_threshold}")

#Seperate Top 10% from remaining 90%
top_10_percent = seller_revenue.head(top_10_percent_threshold)
bottom_90_percent = seller_revenue.tail(total_sellers - top_10_percent_threshold)

#Calculate revenue concentration
total_revenue = seller_revenue['total_revenue'].sum()
top_10_revenue = top_10_percent['total_revenue'].sum()
bottom_90_revenue = bottom_90_percent['total_revenue'].sum()

top_10_percentage = (top_10_revenue / total_revenue) * 100
bottom_90_percentage = (bottom_90_revenue / total_revenue) * 100

print(f"\nRevenue Concentration Analysis")
print("=" * 45)
print(f"Top 10% sellers ({top_10_percent_threshold} sellers):")
print(f"Revenue: R${top_10_revenue:,.2f}({top_10_percentage:.2f}% of total)")
print(f"\nBottom 90% sellers ({total_sellers - top_10_percent_threshold} sellers):")
print(f"Revenue: R${bottom_90_revenue:,.2f}({bottom_90_percentage:.2f}% of total)")
```

```python
# Break down revenue concentration into quartiles
print("Revenue Distribution by Seller Quartile")
print("=" * 45)

quartiles = [0.01, 0.05, 0.10, 0.25, 0.50]

for q in quartiles:
    threshold = int(total_sellers * q)
    revenue = seller_revenue.head(threshold)['total_revenue'].sum()
    percentage = (revenue / total_revenue) * 100
    print(f"Top {int(q*100)}% ({threshold} sellers): "
          f"R${revenue:,.2f} ({percentage:.2f}% of total revenue)")
```

```python
# Identify characteristics of bottom 50% sellers
bottom_50_percent = seller_revenue.tail(int(total_sellers * 0.50))

print("Bottom 50% Seller Analysis")
print("=" * 45)
print(f"Number of sellers:{len(bottom_50_percent)}")
print(f"Average revenue per seller: R${bottom_50_percent['total_revenue'].mean():,.2f}")
print(f"Median revenue per seller: R${bottom_50_percent['total_revenue'].median():,.2f}")
print(f"Lowest revenue seller: R${bottom_50_percent['total_revenue'].min():,.2f}")

# Revenue Per Seller Comparison
top_avg = seller_revenue.head(int(total_sellers * 0.10))['total_revenue'].mean()
bottom_avg = bottom_50_percent['total_revenue'].mean()
multiplier = top_avg / bottom_avg

print(f"\nTop 10% average revenue per seller: R${top_avg:,.2f}")
print(f"Bottom 50% average revenue per seller: R${bottom_avg:,.2f}")
print(f"Top sellers generate {multiplier:.0f}x more revenue than bottom sellers")
```
